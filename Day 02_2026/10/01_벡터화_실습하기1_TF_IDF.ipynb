## 라이브러리 소개

`Scikit-learn`에서는 TF-IDF를 쉽게 진행할 수 있는 툴을 제공하고 있습니다.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer, TfidfVectorizer
```

먼저 `TfidfTransformer` 사용 방법을 살펴보겠습니다.  

`TfidfTransformer`는 이미 단어 카운트 행렬이 있는 경우 이것을 TF-IDF 행렬로 변환해줍니다.  
여기에서는 `CountVectorizer`로 만들고 `TfidfTransformer`를 써보겠습니다.

```python
corpus = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?',
]
```

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)
```

다음은 단어사전입니다.

```python
vectorizer.get_feature_names_out()
```

corpus의 각 문서는 위 단어 사전을 기반으로 문서 벡터가 됩니다.  
예를 들어 단어사전이 ['and', 'document', 'first', 'is', 'one', 'second', 'the', 'third','this']이고 어떤 문서의 벡터값이 [0, 2, 0, 1, 0, 1, 1, 0, 1]이라면 해당 문서에 'document'는 2회, 'is', 'second', 'the', 'this'는 각각 1회 등장했다는 의미가 됩니다.

```python
X.toarray()
```

위 단어 카운트 행렬을 `TfidfTransformer`을 통해 TF-IDF 행렬로 바꿔보겠습니다.

```python
transformer = TfidfTransformer()
tfidf = transformer.fit_transform(X)
tfidf
```

```python
tfidf.toarray()
```

TF-IDF를 계산하는 공식은 다음과 같습니다.

$$\text{TF-IDF} = TF \times IDF$$

여기에서 TF와 IDF는 다음과 같습니다.  

TF(Term Frequency): 특정 단어(t)가 해당 문서(d)에서 얼마나 자주 등장하는가?

$$TF(t,d)=\frac{\text{특정 단어가 등장한 문서}}{\text{전체 문서의 수}}$$

IDF는 다음과 같습니다.

IDF(Inverse Document Frequency): 특정 단어(t)가 모든 문서에 얼마나 희귀한가?

$$IDF(t) = log\frac{전체 문서의 수}{\text{특정 단어가 등장한 문서} + 1}$$

TF-IDF는 위와 같은 계산을 통해 전체적으로 자주 사용되지 않는데 특정 문서에서 많이 사용되면 더 높은 점수를 부가하게 됩니다. 이를 통해 각 단어가 각 문서에서 얼만큼 중요한지 반영하거나 각 문서의 핵심 단어를 추려낼 수 있습니다.

예를 들어 'This is the first document.' 이 문서에서 'is'와 'first'는 둘 다 1회씩 등장하지만 'is'의 경우 다른 문서에서 자주 등장했기 때문에 상대적으로 중요도가 낮게 측정되고 first는 높게 측정됩니다.

위에서는 이해를 위해 단계를 나누어 구현했지만 `TfidfVectorizer`를 통해 간단히 구현할 수 있습니다.

```python
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)
vectorizer.get_feature_names_out()
```

```python
X.toarray()
```

TF-IDF 값에 따른 의미
|     TF-IDF 값 | 의미                                            |
| -----------: | --------------------------------------------- |
|          `0` | 해당 문서에 단어가 없거나 TF-IDF 값이 0                    |
| `0.01 ~ 0.1` | 해당 문서에서 중요도가 낮은 편                             |
|  `0.1 ~ 0.3` | 어느 정도 특징을 나타내는 단어                             |
|     `0.3 이상` | 해당 문서에서 상대적으로 중요한 단어일 가능성이 높음                 |
|       매우 큰 값 | 해당 문서에는 자주 등장하면서 다른 문서에는 상대적으로 드문 단어일 가능성이 높음 |


왜 숫자가 커질까?

TF-IDF는 크게 두 가지를 봅니다.

① TF (Term Frequency)
→ 이 문서에서 그 단어가 얼마나 자주 등장하는가?

② IDF (Inverse Document Frequency)
→ 전체 문서에서 얼마나 드문 단어인가?

그래서:

한 문서에서는 자주 나오지만 전체 문서에서는 드문 단어 → 높은 TF-IDF

```python
#"OO_Document Frequency"는 단지 예시를 위해 사용했던 것이므로 이제 삭제
tfidf_df = tfidf_df.drop('00_Document Frequency', errors='ignore')

#단어들이 열이 아닌 행에 위치하도록
tfidf_df.stack().reset_index()
```

```python
#각 대통령별로 TF-IDF 점수가 가장 높은 단어를 보여주는 히트맵을 만들고
#관심 있는 두 단어인 "전쟁"과 "평화" 옆에 빨간 점을 찍어 보겠습니다.

import altair as alt
import numpy as np

# Terms in this list will get a red dot in the visualization
term_list = ['war', 'peace']

# adding a little randomness to break ties in term ranking
top_tfidf_plusRand = top_tfidf.copy()
top_tfidf_plusRand['tfidf'] = top_tfidf_plusRand['tfidf'] + np.random.rand(top_tfidf.shape[0])*0.0001

# base for all visualizations, with rank calculation
base = alt.Chart(top_tfidf_plusRand).encode(
    x = 'rank:O',
    y = 'document:N'
).transform_window(
    rank = "rank()",
    sort = [alt.SortField("tfidf", order="descending")],
    groupby = ["document"],
)

# heatmap specification
heatmap = base.mark_rect().encode(
    color = 'tfidf:Q'
)

# red circle over terms in above list
circle = base.mark_circle(size=100).encode(
    color = alt.condition(
        alt.FieldOneOfPredicate(field='term', oneOf=term_list),
        alt.value('red'),
        alt.value('#FFFFFF00')
    )
)

# text labels, white for darker heatmap colors
text = base.mark_text(baseline='middle').encode(
    text = 'term:N',
    color = alt.condition(alt.datum.tfidf >= 0.23, alt.value('white'), alt.value('black'))
)

# display the three superimposed visualizations
(heatmap + circle + text).properties(width = 600)
```
